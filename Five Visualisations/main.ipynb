```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
df = pd.read_csv("employee.csv")

print(df)

saving_dir = Path("figures")
saving_dir.mkdir(parents=True, exist_ok=True)

#print("Result Figures will be saved in:", saving_dir)
```

```python
required_columns = {"id", "department", "salary"}

if not required_columns.issubset(df.columns):
    raise ValueError("Check the column names printed in Cell 2.")

df["id"] = pd.to_numeric(df["id"], errors="coerce")
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")

df = df.dropna(subset=["id", "department", "salary"])

colors = {
    "Engineering": "red",
    "HR": "blue",
    "Sales": "green",
    "Finance": "yellow",
    "default": "orange"
}

print("Cleaned dataset:")
print(df.head())
print("Number of rows:", len(df))
```

```python
avg_salary = df.groupby("department")["salary"].mean()

fig, ax = plt.subplots(figsize=(8, 5))

bar_colors = [
    colors.get(dept, colors["default"])
    for dept in avg_salary.index
]

bars = ax.bar(avg_salary.index, avg_salary.values, color=bar_colors)

ax.set_title("Average Salary by Department")
ax.set_xlabel("Department")
ax.set_ylabel("Average Salary (CSV units)")
ax.bar_label(bars, fmt="%.2f")

highest_dept = avg_salary.idxmax()

fig.text(
    0.1, 0.01,
    f"Insight: {highest_dept} has the highest average salary.",
    fontsize=10
)

fig.tight_layout(rect=[0, 0.07, 1, 0.95])
fig.savefig(saving_dir / "bar_chart.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close(fig)
```

**Insight:** The bar chart helps compare employee values across different categories.


```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(df["salary"], bins=5, color="red", edgecolor="black")

ax.set_title("Distribution of Employee Salaries")
ax.set_xlabel("Salary (CSV units)")
ax.set_ylabel("Number of Employees")

fig.text(
    0.1, 0.01,
    "Insight: The histogram shows how employees are distributed across five salary ranges.",
    fontsize=10
)

fig.tight_layout(rect=[0, 0.07, 1, 0.95])
fig.savefig(saving_dir / "histogram.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close(fig)
```

Histogram

Insight: The histogram shows how employee ages are distributed across different ranges.

```python
fig, ax = plt.subplots(figsize=(8, 5))

for dept, group in df.groupby("department"):
    ax.scatter(
        group["id"],
        group["salary"],
        label=dept,
        color=colors.get(dept, colors["default"]),
        s=70
    )

ax.set_title("Employee ID vs Salary by Department")
ax.set_xlabel("Employee ID")
ax.set_ylabel("Salary (CSV units)")
ax.legend(title="Department")

fig.text(
    0.1, 0.01,
    "Insight: Each point represents an employee; colours identify their departments.",
    fontsize=10
)

fig.tight_layout(rect=[0, 0.07, 1, 0.95])
fig.savefig(saving_dir / "scatter_plot.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close(fig)
```

Scatter plot

Insight: The scatter plot helps identify the relationship between the two numerical variables.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12, 15, 14, 18, 22, 25]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    months,
    sales,
    color="orange",
    marker="o",
    linewidth=2,
    label="Monthly Sales"
)

ax.set_title("Monthly Sales Trend")
ax.set_xlabel("Month")
ax.set_ylabel("Sales (units)")
ax.legend()

fig.text(
    0.1, 0.01,
    f"Insight: Sales increased from {sales[0]} units in January to {sales[-1]} units in June.",
    fontsize=10
)

fig.tight_layout(rect=[0, 0.07, 1, 0.95])
fig.savefig(saving_dir / "line_chart.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close(fig)
```

Line chart

Insight: The line chart compares employee values in sorted order.

```python
departments = sorted(df["department"].unique())

salary_data = [
    df.loc[df["department"] == dept, "salary"].values
    for dept in departments
]

fig, ax = plt.subplots(figsize=(8, 5))

box = ax.boxplot(
    salary_data,
    tick_labels=departments,
    patch_artist=True
)

for patch in box["boxes"]:
    patch.set_facecolor("blue")

ax.set_title("Salary Distribution by Department")
ax.set_xlabel("Department")
ax.set_ylabel("Salary (CSV units)")

fig.text(
    0.1, 0.01,
    "Insight: The box plots compare salary spread and help identify possible outliers.",
    fontsize=10
)

fig.tight_layout()
fig.savefig(saving_dir / "box_plot.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close(fig)
```

Box plot

Insight: The box plot shows the median, spread, and possible outliers in the data.
